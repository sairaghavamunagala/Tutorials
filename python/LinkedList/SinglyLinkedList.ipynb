```python
# Singly Linked Lists

class SinglyNode:

  def __init__(self, val, next=None)->None:
    self.val = val
    self.next = next

  def __str__(self)->str:
    return str(self.val)
```

```python
# HEAD->A->B->C->None
HEAD=SinglyNode(1)
A=SinglyNode(3)
B=SinglyNode(5)
C=SinglyNode(9)

HEAD.next=A
A.next=B
B.next=C
print(HEAD)
```

```python
# Traverse the list - O(n)
curr=HEAD
while curr:
  print(curr)
  curr=curr.next
```

```python
# Display the LinkedList
curr=HEAD
elements=[]
while curr:
  elements.append(str(curr.val))
  curr=curr.next
print("->".join(elements))
```

```python

# Search for element
def search(head,val):
  curr=head
  while curr:
    if val==curr.val:
        return True
    curr=curr.next
  return False

print(search(HEAD,5))
```
